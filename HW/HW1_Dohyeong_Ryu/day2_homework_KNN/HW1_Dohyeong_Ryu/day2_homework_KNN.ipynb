# HW1 - KNN from Scratch (MNIST)

## 1. Load Data
- `MNIST_training.csv`: 949 samples (classes 0-9, ~95 each)
- `MNIST_test.csv`: 50 samples (5 per class)
- Each row: column 0 = label, columns 1-784 = 28x28 pixel values (0-255)

```python
import numpy as np


def load_data(path):
    data = np.loadtxt(path, delimiter=",", skiprows=1)
    y = data[:, 0].astype(int)  
    X = data[:, 1:]              
    return X, y


X_train, y_train = load_data("MNIST_training.csv")
X_test, y_test = load_data("MNIST_test.csv")

print("train:", X_train.shape, "/ test:", X_test.shape)
```

```python
neighbors = np.zeros((len(X_test), len(X_train)), dtype=int)
for i in range(len(X_test)):
    d = np.sqrt(((X_train - X_test[i]) ** 2).sum(axis=1))  
    neighbors[i] = np.argsort(d) 

print("neighbors shape:", neighbors.shape)
```

```python
#Euclidean distance
from collections import Counter

print("K   Accuracy")
for K in range(1, 22, 2):
    correct = 0
    for i in range(len(X_test)):
        knn_labels = y_train[neighbors[i, :K]]         
        pred = Counter(knn_labels).most_common(1)[0][0] 
        if pred == y_test[i]:
            correct += 1
    acc = correct / len(X_test)
    print(f"{K:2d}   {acc:.2f}")
```

```python
# Manhatten distance
neighbors_man = np.zeros((len(X_test), len(X_train)), dtype=int)
for i in range(len(X_test)):
    d = np.abs(X_train - X_test[i]).sum(axis=1)   
    neighbors_man[i] = np.argsort(d)            

print("== Manhattan ==")
print("K   Accuracy")
for K in range(1, 22, 2):
    correct = 0
    for i in range(len(X_test)):
        knn_labels = y_train[neighbors_man[i, :K]]
        pred = Counter(knn_labels).most_common(1)[0][0]
        if pred == y_test[i]:
            correct += 1
    print(f"{K:2d}   {correct/len(X_test):.2f}")
```

```python
# Cosine
neighbors_cos = np.zeros((len(X_test), len(X_train)), dtype=int)
for i in range(len(X_test)):
    sim = (X_train @ X_test[i]) / (np.linalg.norm(X_train, axis=1) * np.linalg.norm(X_test[i]))
    neighbors_cos[i] = np.argsort(-sim)           

print("== Cosine ==")
print("K   Accuracy")
for K in range(1, 22, 2):
    correct = 0
    for i in range(len(X_test)):
        knn_labels = y_train[neighbors_cos[i, :K]]
        pred = Counter(knn_labels).most_common(1)[0][0]
        if pred == y_test[i]:
            correct += 1
    print(f"{K:2d}   {correct/len(X_test):.2f}")
```
